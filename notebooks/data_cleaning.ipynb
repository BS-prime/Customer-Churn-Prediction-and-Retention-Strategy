## Cleaning the dataset

```python
# change the str datatype of each values
str_cols = df.select_dtypes(include='str').columns
df[str_cols] = df[str_cols].astype(object)
df.dtypes
```

```python
# replace the null values
df["Total Charges"] = df["Total Charges"].replace(" ", np.nan)
df["Total Charges"] = df["Total Charges"].astype(float)
df.dtypes
```

```python
# check number of unique value of each features
df.nunique().sort_values()
```

- Count, Country, State has just one unique values so they are not a good predictor.
- CustomerID is also irrelevant while making prediction.
- Having latitude and longitude values in churn prediction model would simply produce noise instead of signal.

```python
# Drop the columns
df = df.drop(columns=["Count", "Country", "State", "CustomerID", "Lat Long", "Latitude", "Longitude"])
df.shape
```

# Save file to parquet

```python
# create directory
FILE_PATH = Path.cwd().parent / "data" / "raw"
FILE_PATH.parent.mkdir(parents=True, exist_ok=True)

# save the cleaned file as csv
df.to_parquet(FILE_PATH / "clean_customer_churn.parquet", index=False, engine="fastparquet")
```
